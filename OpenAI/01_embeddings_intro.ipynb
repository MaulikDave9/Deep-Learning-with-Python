#### Getting standard embedding from `text-embedding-3-small` model

```python
#!pip install openai
```

```python
import os
from dotenv import load_dotenv

# Load environment variables from the .env file in your directory
load_dotenv()

# Retrieve the API key from the environment
api_key = os.getenv("OPENAI_API_KEY")

# Verify if the key was loaded successfully
if api_key:
    print(f"✓ API Key loaded successfully. Starts with: {api_key[:7]}...")
else:
    print("✗ API Key not found. Please check your .env file.")
```

```python
from openai import OpenAI
client = OpenAI()
response = client.embeddings.create(
    input="All wheels on the bus go round and round.",
    model="text-embedding-3-small"
)
print(response)
```

```python
print(len(response.data[0].embedding))
```

#### Shortening the embeddings

```python
from openai import OpenAI
client = OpenAI()
response = client.embeddings.create(
    input="All wheels on the bus go round and round.",
    model="text-embedding-3-small",
    dimensions=256
)
print(len(response.data[0].embedding))
```
